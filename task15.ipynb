```python
import numpy as np

# Student marks dataset
marks = np.array([78, 85, 92, 67, 88, 74, 95, 81, 69, 90])

# Statistical calculations
mean = np.mean(marks)
median = np.median(marks)
std_dev = np.std(marks)
minimum = np.min(marks)
maximum = np.max(marks)

# Display results
print("Student Marks:", marks)
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std_dev)
print("Minimum Marks:", minimum)
print("Maximum Marks:", maximum)

# Compare mean and median
print("\nComparison:")
if mean > median:
    print("Mean is greater than median.")
elif mean < median:
    print("Median is greater than mean.")
else:
    print("Mean and median are equal.")

# Find unusually high and low values
print("\nObservations:")

if maximum > mean + std_dev:
    print("Maximum value is unusually high.")
else:
    print("Maximum value is not unusually high.")

if minimum < mean - std_dev:
    print("Minimum value is unusually low.")
else:
    print("Minimum value is not unusually low.")
```
